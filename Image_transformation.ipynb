#### Common Errors when You Laod a Image

The error indicates that OpenCV is attempting to display an image using cv2.imshow, but the image provided has invalid dimensions—its width or height is 0. This typically happens when:

**1. The image file was not found:** The path to the image file is incorrect or the file does not exist.

**2. The image file is corrupt:** OpenCV could not decode the file properly.

**3. Incorrect OpenCV image loading:** OpenCV failed to load the image due to unsupported formats or improper file permissions.

### Cente the image window

```python
import cv2
from matplotlib import pyplot as plt

# Load the image
image_path = r"images\flower.jpg"
image = cv2.imread(image_path)

if image is None:
    print(f"Error: unable to load the image.")
else:
    plt.imshow(image)
    
    screen_width = 1920
    screen_height = 1080

    # Get Image dimensions
    image_height, image_width = image.shape[:2]
    print(f"Image_width: {image_width}, Image_height: {image_height}")
    
    # Calculat top-left corner position for centering
    x_pos = (screen_width - image_width) // 2 
    y_pos = (screen_height - image_height) // 2

    cv2.namedWindow("Centered Image", cv2.WINDOW_NORMAL)
    cv2.moveWindow("Centered Image", x_pos, y_pos)
    cv2.imshow("Centered Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

#### Dynamically Resize and centre the image

```python
import cv2
import pyautogui

image = cv2.imread(r"Images\flower.jpg")
if image is None:
    print("Error: Unable to load the image " )
else:
    scrren_width, screen_height = pyautogui.size()
    image_height, image_width = image.shape[:2]

    # Check wheth image is larger than the screen
    if image_width > screen_width or image_height > screen_height:
        # Calculat scale to fit the image within the screen
        scale = min(scrren_width/image_width, screen_height/image_height)
        new_width = image_width * scale
        new_height = image_height * scale

        # Resize the image
        image =  cv2.resize(image, (new_width, new_height))
    
    # calculate center position of window
    x_pos = (screen_width - image.shape[1]) //2
    y_pos = (screen_height - image.shape[0])//2

    cv2.namedWindow("Centered Image", cv2.WINDOW_NORMAL)
    cv2.moveWindow("Centered Image", x_pos, y_pos)
    cv2.imshow("Centered Image", image)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
