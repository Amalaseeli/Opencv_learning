### Histogram Calculation in OpenCV

Histograms provide a graphical representation of the distribution of pixel intensity values in an image. They are essential for understanding the contrast, brightness, and intensity spread of an image. Below are examples demonstrating histogram analysis using OpenCV.

cv2.calcHist(images, channels, mask, histSize, ranges[, hist[, accumulate]])

1. **images :** it is the source image of type uint8 or float32. it should be given in square brackets, ie, "[img]".
    
2. **channels :** it is also given in square brackets. It is the index of channel for which we calculate histogram. For example, if input is grayscale image, its value is [0]. For color image, you can pass [0], [1] or [2] to calculate histogram of blue, green or red channel respectively.
    
3. **mask :** mask image. To find histogram of full image, it is given as "None". But if you want to find histogram of particular region of image, you have to create a mask image for that and give it as mask. (I will show an example later.)
    
4. **histSize :** this represents our BIN count. Need to be given in square brackets. For full scale, we pass [256].
    
5. **ranges :** this is our RANGE. Normally, it is [0,256].

#### Histogram equalization

Histogram equalization enhances contrast by redistributing pixel intensity values.
It is useful for improving the visibility of details in images with poor contrast.

```python
image = cv2.imread("Images/peacock.jpg", cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found")
    exit()

# Apply histogram equalization
equalized_image = cv2.equalizeHist(image)

# Calculate histograms for both images
original_hist = cv2.calcHist([image], [0], None, [256], [0,256])
equalized_hist = cv2.calcHist([equalized_image], [0], None, [256], [0,256])

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Original Histogram")
plt.plot(original_hist, color="black")
plt.xlim([0, 256])


plt.title("Equalized Histogram")
plt.subplot(1,2,2)
plt.plot(equalized_hist, color = "black")
plt.xlim([0, 256])

plt.show()
```

```python
image = cv2.imread("Images/peacock.jpg", cv2.IMREAD_GRAYSCALE)

# Check if the image was loaded
if image is None:
    print("Error: Image not found.")
    exit()

# Apply histogram equalization
equalized_image = cv2.equalizeHist(image)

# Calculate histograms for both original and equalized images
original_hist = cv2.calcHist([image], [0], None, [256], [0, 256])
equalized_hist = cv2.calcHist([equalized_image], [0], None, [256], [0, 256])

# Plot the results using subplots
plt.figure(figsize=(12, 8))

# Display original image
plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap="gray")
plt.axis("off")

# Display histogram of original image
plt.subplot(2, 2, 2)
plt.title("Original Histogram")
plt.plot(original_hist, color="black")
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# Display equalized image
plt.subplot(2, 2, 3)
plt.title("Equalized Image")
plt.imshow(equalized_image, cmap="gray")
plt.axis("off")

# Display histogram of equalized image
plt.subplot(2, 2, 4)
plt.title("Equalized Histogram")
plt.plot(equalized_hist, color="black")
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# Show the plots
plt.tight_layout()
plt.show()
```

#### Cumulative Histogram

A cumulative histogram shows the running total of pixel frequencies, helping analyze global intensity distribution.

```python
import numpy as np

image = cv2.imread("Images/peacock.jpg", cv2.IMREAD_GRAYSCALE)
if image is None:
    print("Error: Image not found")
    exit()

equalized_image = cv2.equalizeHist(image)

# Calculate histogram
original_hist = cv2.calcHist([image], [0], None, [256], [0,256])
equalized_hist = cv2.calcHist([equalized_image], [0], None, [256], [0, 256])

# calculate cumulative histogram
cumulative_original = np.cumsum(original_hist)
cumulative_equalized = np.cumsum(equalized_hist)

# Normalize cumulative histograms for better visualization
cumulative_original = cumulative_original / cumulative_original.max()
cumulative_equalized = cumulative_equalized / cumulative_equalized.max()

# Plot images and cumulative histograms
plt.figure(figsize=(12, 8))

# Original image
plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap="gray")
plt.axis("off")

# Cumulative histogram of original image
plt.subplot(2, 2, 2)
plt.title("Cumulative Histogram (Original)")
plt.plot(cumulative_original, color="blue")
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Cumulative Frequency")

# Equalized image
plt.subplot(2, 2, 3)
plt.title("Equalized Image")
plt.imshow(equalized_image, cmap="gray")
plt.axis("off")

# Cumulative histogram of equalized image
plt.subplot(2, 2, 4)
plt.title("Cumulative Histogram (Equalized)")
plt.plot(cumulative_equalized, color="green")
plt.xlim([0, 256])
plt.xlabel("Pixel Intensity")
plt.ylabel("Cumulative Frequency")

# Adjust layout for better visualization
plt.tight_layout()
plt.show()
```
