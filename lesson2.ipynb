### Individual Channels in RGB Images

```python
image = cv2.imread('Images/large.jpg')
cv2.imshow("Image", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
B, G, R = cv2.split(image)
cv2.imshow("Red", R)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Green", G)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Blue", B)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Merge

```python
import cv2
merged = cv2.merge([B,G, R ])
cv2.imshow("Merged", merged)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Lets amplify the blue color

```python
merged = cv2.merge([B+100, G, R+200])
cv2.imshow("Merged", merged)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### HSV (Hue, Saturation, and Value) Images
### H: 0 - 180, S: 0 - 255, V: 0 - 255



```python
image = cv2.imread("Images/Tiger.jpg")
cv2.imshow("Tiger", image)

hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
cv2.imshow("HSv", hsv_image)

cv2.waitKey()
cv2.destroyAllWindows()
```

```python
cv2.imshow("hue chanel", hsv_image[0:, 0 :, 0])
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Converting image into gray
# and resize in the image
cv2.imshow('Saturation channel', hsv_image[0:,0 :, 1])
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Converting image into black and white
# and resize in the image
cv2.imshow('Value channel', hsv_image[0:,0 :, 2])
cv2.waitKey()
cv2.destroyAllWindows()
```

### 2.7 Analysing Images using openCV Functions

```python
# Let's look at the individual color levels for the first pixel (0,0)
import cv2
#import numpy as np

path = r'Images\clouds.jpg'
image = cv2.imread(path)

cv2.imshow('Clouds',image)

cv2.waitKey()
cv2.destroyAllWindows()
```

```python
print(image.shape)
```

```python
print("Height of the image", int(image.shape[0]), 'pixels')
print("Width of the image", int(image.shape[1]), 'pixels')
```

```python
print(image.shape[:])
print(image.shape[:1])
print(image.shape[:2])
print(image.shape[1:])
print(image.shape[2:])
```

```python
B, G , R = image[0,0]
cv2.imshow("image", image[0,0])
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# BGR Values for the first 0,0 pixel
B, G, R = image[0,0] 
print("Blue Value: ", B)
print("Green Value: ", G)
print("Red Value: ", R)
```

```python
# BGR Values for the last 829,1244 pixel
B, G, R = image[829,900] 
print("Blue Value: ", B)
print("Green Value: ", G)
print("Red Value: ", R)
```
